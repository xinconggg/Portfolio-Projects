```python
from pathlib import Path
import sys
```

## Configuration

```python
PROJECT_ROOT = Path.cwd().parents[2]
sys.path.insert(0, str(PROJECT_ROOT))

from src.data.ingestion import load_raw_option_chain

DATA_PATH = PROJECT_ROOT / "src" / "data" / "raw" / "SPY" / "2010" / "spy_eod_201001.txt"
```

## Load

```python
df = load_raw_option_chain(DATA_PATH)
```

## Basic Inspection

```python
print("=" * 80)
print("INGESTION CHECK")
print("=" * 80)

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nDtypes:")
print(df.dtypes)

print("\nFirst 5 rows:")
print(df.head())

print("\nMissing values:")
print(df.isna().sum())

print("\nDate range:")
print("Quote:", df["quote_date"].min(), "→", df["quote_date"].max())
print("Expiry:", df["expire_date"].min(), "→", df["expire_date"].max())

print("\nUnderlying:")
print(df["underlying_last"].describe())

print("\nStrike:")
print(df["strike"].describe())

print("\nDTE:")
print(df["dte"].describe())

print("\nVolume missingness:")

print(
    "Call volume missing:",
    df["c_volume"].isna().sum()
)

print(
    "Call volume zero:",
    (df["c_volume"] == 0).sum()
)

print(
    "Put volume missing:",
    df["p_volume"].isna().sum()
)

print(
    "Put volume zero:",
    (df["p_volume"] == 0).sum()
)
```
